```python
from mastodoner import Crawler

crawler = Crawler()

# Wissenschaftliche Instanz crawlen
posts = crawler.instance_public_timeline('sigmoid.social', limit=100)

import json
with open('testdaten.jsonl', 'w') as f:
    for post in posts:
        f.write(json.dumps(post) + '\n')
```
