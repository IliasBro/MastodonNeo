```python
from mastodoner import Crawler
import json

crawler = Crawler()

# Timeline crawlen
posts = crawler.instance_timeline('sigmoid.social', max_limit=100)

# In Datei speichern
with open('Data/SampleData.jsonl', 'w') as f:
    for post in posts:
        f.write(json.dumps(post) + '\n')

print(f"{len(posts)} Posts gespeichert")
```
